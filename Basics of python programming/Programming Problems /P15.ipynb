# Wap to calculate salary of an employee given his base pay(to be entered by the user),HAR = 10 per cent of basic pay , TA =5 per cent of basic pay . Define HAR and TA as constant and use them to calculate the salary of employee

```python
base_sal = float(input("Enter the base salary :"))
Har = 0.10*base_sal
Ta = 0.05*base_sal
Total_sal = base_sal+Har+Ta
print("total employee salary is :=",Total_sal)
```
