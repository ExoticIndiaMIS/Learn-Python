# Arithmetic operation with float point number

```python
num1 = float(input("enter the first number :"))
num2 = float(input("enter the second number :"))
add = num1+num2
sub = num1-num2
div = num1/num2
mult = num1*num2
modul = num1%num2
floor_div = num1//num2
expon = num1**num2
print("addition :",add)
print("subtraction :",sub)
print("division :",div)
print("multiplication :",mult)
print("modulus :",modul)
print("Floor division :",floor_div)
print("exponent :",expon)
```
