# Wap to calculate the bill amount for item given its quantity sold,value,discounts and tax.

```python
qty = int(input("enter the quantity of item sold  ="))
val = float(input("enter the value of item ="))
discount = float(input("enter the discount on item in percentage ="))
tax = float(input("enter the tax on the item ="))
amount = qty*val
discount_amt = (amount*discount)/100
sub_total = amount - discount_amt
tax_amt = (sub_total*tax)/100
total_amt = sub_total+tax_amt
print("\n")
print("**********Bill**********")
print("quantity sold : \t" , qty)
print("price per item : \t",val)
print("\n\t\t -----------------")
print("Amount: \t\t ",amount)
print("Discount: \t\t ",discount_amt)
print(" \n \t\t -----------------")
print("Discount total: \t ",sub_total)
print("Tax: \t\t\t ",tax_amt)
print("\n\t\t -----------------")
print("total amount to be pade : ",total_amt)
```
