# Wap to calculate the total amout of money in the piggybank

```python
coin_10 = int(input("enter the number of 10 coin ="))
coin_5 = int(input("enter the number of 5 coin ="))
coin_2 = int(input("enter the number of 2 coin ="))
coin_1 = int(input("enter the number of 1 coin ="))
total_amount = coin_10*10+coin_5*5+coin_2*2+coin_1*1
print("total amount of coin =",total_amount)
```
