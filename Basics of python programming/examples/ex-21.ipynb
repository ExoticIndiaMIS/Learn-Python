# Wap to perform addition subtraction multiplication division modulo division on two integer

```python
num1 = int(input("enter the num1 "))
num2 = int(input("enter the num2 "))
add = num1+num2
sub = num1-num2
mult = num1*num2
divid = num1/num2
md = (num1%num2)
print("add=",add)
print("sub=",sub)
print("multiplication=",mult)
print("divide=",divid)
print("modulo divide =",md)
```
