# WAP to enter a number and display its hex and octal equivalent and its square root

```python
# Input a number from the user
num = int(input("Enter a number: "))

# Convert to hexadecimal and octal (avoid using 'hex' or 'oct' as variable names)
value1= hex(num)
value2= oct(num)

# Calculate square root manually using exponentiation
sqrt_value = num ** 0.5

# Display the results
print(f"Hexadecimal: {value1}")
print(f"Octal: {value2}")
print(f"Square Root: {sqrt_value:.2f}")
```
