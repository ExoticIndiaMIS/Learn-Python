# Ac company decides to give bonus to all employees on diwali. A 5% bonus on salary is give to male workers and 10% to femal workers. Wap to enter the salary of employee and sex of the employees

```python
sex = input("enter the sex of the employee (m or f):=")
sal = int(input("enter the salary of the employee :="))
if (sex == 'm'):
    bonus = 0.05*sal
else:
    bonus = 0.10*sal
amount = sal +bonus
print("Salary = ",sal)
print("Bonus = ",bonus)
print("***************************")
print("Amount to be pade =",amount)
```
