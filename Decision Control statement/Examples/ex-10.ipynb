# Wap to prompt the user to enter a number and then print its interval

```python
num = int(input("enter the number between 1 to 30="))
if (num>=0 and num<10):
    print("number is in between 0 to 10 ")
if (num>=10 and num<20):
    print("number is in between 10 to 20")
if (num>=20 and num<30):
    print("number is in between 20 to 30")
```
