# Wap to take input from the user and then check whether it is a number or character . if it is a character ,determine whether it is uppercase or lowercase.

```python
ch = input("enter the character=")
if(ch>='A' and ch<='Z'):
    print("entered character is in uppercase ")
elif(ch>='a' and ch<='z'):
    print("entered character is in lowercase")
else :
    print("entered character is a number")
```
