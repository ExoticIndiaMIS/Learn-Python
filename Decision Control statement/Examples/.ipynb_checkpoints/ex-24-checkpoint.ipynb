# Wap to read tghe number until-1 is encountered . Also count the negative , positive and zeroes by the user

```python
print("enter -1 to exit")
sumn=0
sump=0
sumz=0
while (1):# it is used for infinite loop
    num = int(input("enter the value of num ="))
    if(num == -1):
        break
    if(num == 0):
        sumz=sumz+1
    elif(num >0):
        sump = sump+1
    else:
        sumn = sumn+1
print("total number of negative number=",sumn)
print("total number of positive number=",sump)
print("total number of zeros =",sumz)
    
```
