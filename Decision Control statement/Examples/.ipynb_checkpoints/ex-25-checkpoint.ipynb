# Wap to read the number until-1 is encountered . Find the average of positive number and negative number entered by the user .

```python
countp=0
sump=0
countn=0
sumn=0
print("enter -1 to exit")
num = 0
while(num!=-1):
    num = int(input("enter the number"))
    if(num<0):
        countn+=1
        sumn +=num
    else:
        countp +=1
        sump +=num
avgp = sump/countp
avgn = sumn/countn
print(avgp)
print(avgn)


   
```

```python
countp = 0
sump = 0
countn = 0
sumn = 0
print("Enter -1 to exit")

num = 0
while num != -1:
    num = int(input("Enter the number: "))
    if num == -1:
        break  # Exit the loop
    if num < 0:
        countn += 1
        sumn += num  # Correct summation for negatives
    elif num > 0:
        countp += 1
        sump += num  # Correct summation for positives

# Calculate averages with checks for zero count
avgp = sump / countp if countp > 0 else 0
avgn = sumn / countn if countn > 0 else 0

print(f"Average of positive numbers: {avgp}")
print(f"Average of negative numbers: {avgn}")
```
