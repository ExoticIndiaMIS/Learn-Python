# Wap that prompts thje user to enter a number between 1-7 and display the crosponding day of the week.

```python
num = int(input("enter the number between 1 to 7 ="))
if(num ==1):
    print("sunday")
elif(num==2):
    print("monday")
elif(num==3):
    print("tuesday")
elif(num ==4):
    print("wednesday")
elif(num ==5):
    print("thursday")
elif(num==6):
    print("friday")
elif(num==7):
    print("saturday")
else:
    print("wrong input")
```
