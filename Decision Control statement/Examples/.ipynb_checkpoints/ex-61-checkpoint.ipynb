# Wap to calculate the square root of a number.demonstrate the use of break and continue statement

```python
import math

while True:
    num = input("Enter a number to find its square root (or type 'exit' to stop): ")

    if num.lower() == 'exit':
        print("Exiting the program.")
        break  
    
    if not num.isdigit():
        print("Invalid input! Please enter a positive number.")
        continue 
    
    num = float(num)
    
    if num < 0:
        print("Square root of a negative number is not defined in real numbers.")
        continue  
    
    sqrt_value = math.sqrt(num)
    print(f"Square root of {num} is {sqrt_value:.2f}")
```
