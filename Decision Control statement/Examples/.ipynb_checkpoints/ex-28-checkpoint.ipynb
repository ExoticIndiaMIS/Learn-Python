# Wap to enter the binary number and convert it into decimel

```python
bin = int(input("enter the binery number"))
dis = 0
i = 0
while(bin !=0):
    remend = bin%10
    dis = dis+remend*(2**i)
    bin = bin//10
    i = i+1
print("decimal number=",dis)
```

# Wap to convert the decimel number to octal 

```python
num = int(input("enter the number+"))
oct = 0 
i = 0
while(num !=0):
    remend = num%8
    oct = oct+remend*(10**i)
    num = num//8
    i = i+1
print("octal =",oct)
```

# Wap to convert the decimel number to hexadecimal

```python
num = int(input("Enter the number: "))
hex_num = ""  # To store the hexadecimal number
hex_digits = "0123456789ABCDEF"  # Hexadecimal digits

i = 0
while num != 0:
    remainder = num % 16  # Find remainder when dividing by 16
    hex_num = hex_digits[remainder] + hex_num  # Append the corresponding hex digit
    num = num // 16  # Integer division by 16 to get the next digit
    i += 1

print("Hexadecimal =", hex_num)
```
