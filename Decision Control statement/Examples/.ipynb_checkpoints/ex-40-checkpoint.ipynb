# Wap to using while loop to read the numbers unti -1 is encounterd . Also count the number of prime and compositive number entered.

```python
tp = 0
tc = 0
while(1):
    n = int(input("enter the number="))
    if(n==-1):
        break
    elif(n==0):
        print("enter the another number")
    elif(n==1):
        print("1 is nither a prime nor composite")
    else:
        is_prime = True
        for i in range(2, n):  
            if n % i == 0:
                is_prime = False
                break
        if is_prime:
            tp = tc+1
            
        else:
            tc = tc+1
print(tp)
print(tc)
            
           
    
```

```python
n = int(input("enter the number="))
if (n==0):
    print("enter the another number")
elif(n==1):
    print("1 is nither a prime nor composite")
else:
    is_prime = True
    for i in range(2, n):  
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{n} is a prime number.")
    else:
        print(f"{n} is a composite number.")
    
    
```
