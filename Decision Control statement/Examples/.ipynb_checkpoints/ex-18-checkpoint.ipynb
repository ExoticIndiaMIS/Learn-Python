# Wap to calculate the root of quadratic equation

```python
a = int(input("enter the value of a"))
b = int(input("enter the value of b"))
c = int(input("enter the value of c"))
d = (b*b)-(4*a*c)
if(d>0):
    print("two real root")
    root1=(-b+d**0.5)/2*a
    root2=(-b-d**0.5)/2*a
    print("Root 1=",root1, "\tRoot 2 =",root2)
elif(d==0):
    print("Equal roots")
    root1 = (-b/2*a)
    print("Root =",root1)
else:
    print("imaginary roots")
```
