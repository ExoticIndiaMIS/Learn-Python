# Wap to read a character untile a * is encountered. Also countthe number of uppercase,lowercase and number entered by the users. 

```python
ch = input("Enter any character: ")
num_con = 0
up_con = 0
low_con = 0

while ch != '*':
    if '0' <= ch <= '9':  # Check if the character is a digit
        num_con += 1
    elif 'a' <= ch <= 'z':  # Check if the character is lowercase
        low_con += 1
    elif 'A' <= ch <= 'Z':  # Check if the character is uppercase
        up_con += 1
    ch = input("Enter the character: ")

# Display the counts
print("Number of digits:", num_con)
print("Number of uppercase letters:", up_con)
print("Number of lowercase letters:", low_con)
```
