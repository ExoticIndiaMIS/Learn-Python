# Wap using for loop to print all the numbers from m-n thereby classify them as odd or even.

```python
m = int(input("enter the value of m ="))
n = int(input("enter the value of n ="))
for i in range(m,n+1):
    if (i%2==0):
        print(i,"even number")
    else:
        print(i,"odd number")
```
