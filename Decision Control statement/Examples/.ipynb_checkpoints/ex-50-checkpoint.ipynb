# Wap to generate a calender of month given the start_day and the number of days in that month.

```python
# Days of the week
days_of_week = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
print(" ".join(days_of_week))
start_day = int(input("Enter the start day of the month (0 for Sun, 1 for Mon, ..., 6 for Sat): "))
num_days = int(input("Enter the number of days in the month: "))

calendar = "    " * start_day  # 4 spaces per day
for day in range(1, num_days + 1):
    calendar += f"{day:3} "  
    if (start_day + day) % 7 == 0:
        calendar += "\n"

print(calendar)
```
