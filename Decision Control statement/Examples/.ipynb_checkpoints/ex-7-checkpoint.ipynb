# Wap to enter any character . if the entered character is in lowercase the convert it in uppercase and vise versa

```python
char = input("Enter the character =")
if (char>='A' and char <='Z'):
    char = char.islower()
    print("entered character is in upper case and .Lowercase character is ",str(char))
else:
    char = char.isupper()
    print("entered character is in lowercase and .uppercase character is " ,str(char))
```
