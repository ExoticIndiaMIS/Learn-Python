# Wap to calculate the GCD of two number

```python
num1 = int(input("enter the number one ="))
num2 = int(input("enter the number two="))
if(num1>num2):
    dividend = num1
    divisor = num2
else:
    dividend = num2
    divisor = num1
while(divisor!=0):
    remender = dividend%divisor
    dividend = divisor
    divisor = remender
print("GCD of number is =",dividend)
```
