# Wap to calculate tax given the following condition :


```python
# if income<150000 no tax
# if income 150001-300000 charge 10%
# if income 300001-500000 charge 20%
# if income is above 500001 charge 30%
```

```python
income = int(input("enter your income ="))
if (income == 150000):
    print("no tax")
elif(income>=150001 and income<=300000):
    print("tax =",income*0.10)
elif(income>=300001 and income<=500000):
    print("tax =",income*0.20)
else :
    print("tax=",income*0.30)
```
