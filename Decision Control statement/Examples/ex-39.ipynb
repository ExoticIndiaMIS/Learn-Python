# Wap to classify a given number as prime or composite.

```python
n = int(input("enter the number="))
if (n==0):
    print("enter the another number")
elif(n==1):
    print("1 is nither a prime nor composite")
else:
    is_prime = True
    for i in range(2, n):  
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{n} is a prime number.")
    else:
        print(f"{n} is a composite number.")
    
    
```
