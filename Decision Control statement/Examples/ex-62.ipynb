# Wap that prompt user to enter numbers. that procss will repeat until user enter-1 Finally, the program print the count of prime and composite number.

```python
c_count = 0
p_count = 0
n = int(input("Enter a number: "))

while n != -1:
    if n < 2:
        print(f"{n} is Neither Prime nor Composite")
    else:
        is_prime = True
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                is_prime = False
                break
        if is_prime:
            p_count += 1
        else:
            c_count += 1
    n = int(input("Enter the new number: "))

print("Total Prime Numbers:", p_count)
print("Total Composite Numbers:", c_count)
```
