# Wap that reads an integers until the user wants to stop . When the user stop entering numbers,display the larger of all the number entered.

```python
largest = None  # Variable to store the largest number

while True:
    num = input("Enter an integer (or type 'stop' to end): ")  
    
    if num.lower() == "stop":  # Check if the user wants to stop
        break  

    num = int(num)  # Convert input to integer

    if largest is None or num > largest:  # Update the largest number
        largest = num  

if largest is not None:
    print("The largest number entered is:", largest)
else:
    print("No numbers were entered.")
```
