# Wap that count the number of lowercase character , uppercase character and digit entered by the user.

```python
u_count = 0
l_count = 0
d_count = 0
while True:
    text = input("Enter a string (or type 'exit' to stop): ")

    if text.lower() == 'exit':  # Stop taking input if the user types 'exit'
        break
    for i in text:
        if i.isupper():
            l_count = l_count+1
        elif i.islower():
            u_count = u_count+1
        else:
            d_count = d_count+1            
print(u_count)
print(l_count)
print(d_count)
```
