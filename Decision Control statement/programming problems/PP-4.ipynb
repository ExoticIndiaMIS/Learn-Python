# WAP  to determine whether a digit,uppercase,or a lower case character was entered

```python
ch = input("enter the character=")
if ch>='0' and ch<='9':
    print("enter character is digit")
elif ch>='a' and ch<='z':
    print("entered character is lower case alphabate")
elif ch>='A' and ch<='Z':
    print("entred character is upper case alphabate")
else:
    print("white space is entered" )
```
