# Wap that prompt the user to enter five words. If the length of any word is less than 6 character then it ask the user to enter it againHower if the word is 6 or more character , then it display it on the screen.

```python
count = 0  
while count < 5: 
    word = input(f"Enter word {count + 1}: ")

    if len(word) >= 6:  
        print(f"Accepted: {word}")
        count += 1  #
    else:
        print("Word must be at least 6 characters long. Try again!")
```
