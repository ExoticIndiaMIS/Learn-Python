# Wap to read an integer. If it is positive then display the corresponding binary representation of that number. The user must enter 999 to stop . In case the user enter negative number, then ignore that input and ask the user to re-enter any different number.

```python
while True:
    num = int(input("Enter a positive integer (999 to stop): "))
    if num == 999:
        print("Program stopped.")
        break  
    if num < 0:
        print("Negative numbers are not allowed. Please enter a positive number.")
        continue  
    binary = bin(num)[2:] 
    print(f"Binary representation of {num} is: {binary}")
```
