# Wap that determines whether a student is eligible for pg course or not . To be eligible, the studen must have obtained more than 80% in X and XII examination , and 70% plus marks in graduation. if the student changes his strem (Science,Commerce,or Arts), then deduct 5% from his graduation score.

```python

x_percent = float(input("Enter X (10th) percentage: "))
xii_percent = float(input("Enter XII (12th) percentage: "))
grad_percent = float(input("Enter Graduation percentage: "))
changed_stream = input("Did you change your stream? (yes/no): ").strip().lower()
if changed_stream == "yes":
    grad_percent -= 5
    print("5% deducted due to stream change.")
if x_percent > 80 and xii_percent > 80 and grad_percent > 70:
    print("Eligible for PG course!")
else:
    print(" Not Eligible for PG course.")
```
