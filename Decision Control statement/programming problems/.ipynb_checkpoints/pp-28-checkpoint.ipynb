# Wap that accept the current date and the date of birth of the user . Then calculate the age of the user and display it on the screen. Note that the date should be displayed in the formate specified as - dd/mm/yy.

```python
c_date = int(input("enter the current date="))
c_month = int(input("enter the current month="))
c_year = int(input("enter the current year="))
print(f"current date is {c_date}/{c_month}/{c_year}")
dob = int(input("enter the date of birth="))
dom = int(input("enter the month of birth="))
doy = int(input("enter the year of birth="))
print(f"date of birth is {dob}/{dom}/{doy}")
age_years = c_year - doy
age_months = c_month - dom
age_days = c_date - dob
if age_days < 0:
    age_months -= 1
    age_days += 30  
if age_months < 0:
    age_years -= 1
    age_months += 12
print("Your age:", age_years, "years,", age_months, "months,", age_days, "days")

```
