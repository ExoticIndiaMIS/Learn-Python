# Wap to implement the following sequence of the numbers.
# 1,8,27,64,......
# -5,-2,0,3,6,9,12,...
# -2,-4,-6,-8,-10,-12,.....
# 1,4,7,10,......

```python
num = 50
for i in range(1,num+1):
    prt = i*i*i
    i = i+1
    print(prt,",",end=" ")
    
```

```python
n = int(input("Enter the number of terms: "))  # User inputs how many terms to generate
term = -5  # First term of the sequence

for i in range(n):
    print(term, end=", " if i < n - 1 else "\n")  # Print with a comma, except for the last term
    term += 3  # Increase by the common difference
```

```python
num = 50
term = -2
for i in range(num):
   print(term, end=", ")  # Print with a comma, except for the last term
   term -= 2
```

```python
n = int(input("Enter the number of terms: "))  # Number of terms for each sequence

# Sequence 1: Cubes of natural numbers
print("\nSequence 1: Cubes of natural numbers")
for i in range(1, n + 1):
    print(i**3, end=", " if i < n else "\n")

# Sequence 2: Arithmetic Progression (-5, -2, 0, 3, ...)
print("\nSequence 2: Arithmetic Progression (-5, -2, 0, ...)")
term = -5
for i in range(n):
    print(term, end=", " if i < n - 1 else "\n")
    term += 3

# Sequence 3: Arithmetic Progression (-2, -4, -6, -8, ...)
print("\nSequence 3: Arithmetic Progression (-2, -4, -6, ...)")
term = -2
for i in range(n):
    print(term, end=", " if i < n - 1 else "\n")
    term -= 2

# Sequence 4: Arithmetic Progression (1, 4, 7, 10, ...)
print("\nSequence 4: Arithmetic Progression (1, 4, 7, 10, ...)")
term = 1
for i in range(n):
    print(term, end=", " if i < n - 1 else "\n")
    term += 3
```
