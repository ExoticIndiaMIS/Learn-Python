# Wap that prompt the user to enter a string. The program calculate and display s the length of the string until the user enter quit.

```python
while (True):
    s = input("enter the string to find length or exit to get out=")
    if s.lower()=='exit':
        print("exiting the loop")
        break
    print(f"The length of the string is {len(s)}")
        
    
    
```
