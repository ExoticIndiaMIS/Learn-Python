# Wap to print the prime factors of a number .

```python
num = int(input("Enter a number: "))

print(f"Prime factors of {num} are: ")
for i in range(2, num + 1):
    while num % i == 0:
        print(i, end=" ")
        num = num // i  
    if num == 1:  
        break
```

```python
num = int(input("Enter a number: "))

print(f"Prime factors of {num} are: ")

# Start with the smallest prime factor (2)
while num % 2 == 0:
    print(2, end=" ")
    num = num // 2

# Check for odd prime factors
i = 3
while i * i <= num:
    while num % i == 0:
        print(i, end=" ")
        num = num // i
    i += 2  # Skip even numbers, as they can't be prime

# If num is still greater than 2, it's a prime number itself
if num > 2:
    print(num)
```
