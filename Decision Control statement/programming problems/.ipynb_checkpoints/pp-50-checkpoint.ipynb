# An employee's total weekly pay is calculated by multiplying the hourly wage and number of hour plus any overtime pay which in turn is calculated as total overtime hours multipled by 1.5 times the hourly wages Write a program that takes as input the hourly wages total regular hour and total overtime hours and print an empoloyee's total weekly pay.

```python
# Get input from the user
hourly_wage = float(input("Enter hourly wage: "))
regular_hours = float(input("Enter total regular hours worked: "))
overtime_hours = float(input("Enter total overtime hours worked: "))

# Calculate regular pay
regular_pay = regular_hours * hourly_wage

# Calculate overtime pay (1.5 times the hourly wage)
overtime_pay = overtime_hours * (1.5 * hourly_wage)

# Calculate total weekly pay
total_weekly_pay = regular_pay + overtime_pay

# Display the total pay
print(f"Total weekly pay: ${total_weekly_pay:.2f}")
```
