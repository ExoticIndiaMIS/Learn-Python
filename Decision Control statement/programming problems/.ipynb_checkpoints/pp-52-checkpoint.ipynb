# Wap that convert grams to kilograms and meters to centimeters.

```python

grams = float(input("Enter weight in grams: "))
kilograms = grams / 1000  # 1 kg = 1000 grams
print(f"{grams} grams = {kilograms} kilograms")
meters = float(input("Enter length in meters: "))
centimeters = meters * 100  # 1 meter = 100 centimeters
print(f"{meters} meters = {centimeters} centimeters")
```
