# Wap that prompt users to enter a numbers. Once the user enter -1 , it displays the count,sum,and average of even numbers and that of odd numbers.

```python
count_e = 0
count_o = 0
sum_e = 0
sum_o = 0

while (True):
    n = int(input("Enter the number = "))
    if n == -1:
        break
    if n % 2 == 0:
        count_e = count_e + 1
        sum_e = sum_e + n
    else:
        count_o = count_o + 1  # Corrected the typo here
        sum_o = sum_o + n

# Calculate averages (avoid division by zero)
even_avg = sum_e / count_e if count_e > 0 else 0
odd_avg = sum_o / count_o if count_o > 0 else 0

print(f"Even Average: {even_avg}")
print(f"Odd Average: {odd_avg}")
```
