# Wap to calculate parking charge of a vehicle. Enter the type of vehical as a character (like c for car and b for bus,etc). and number of hour, then calculate charge as given below :
# truck/bus-20$perhour
# car-10$ per hour
# scooter/cycle/motor cycle-5$per hour

```python
print("enter c for car b or t for truck and m for 2 weller vehical ")
type = input("enter the vehical type=")
hour = int(input("enter the hour you want to keep ="))
price = 0
if type=='t' or type == 'b':
    price = hour*20
elif type=='c':
    price = hour*10
elif type =='m':
    price = hour*5
else:
    print("enter the correcte type of vehical")
print(f"for vehical type {type} parking charge is {price}")
```
