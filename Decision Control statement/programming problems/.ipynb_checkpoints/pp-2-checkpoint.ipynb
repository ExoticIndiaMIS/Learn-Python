# Wap to prompt user to enter a character(O,A,B,C,F).Then usine if-elif-else construct display Outstanding,Very Good , Good, Average and Fail respectively.

```python
ch = input("enter a character(O,A,B,C,F)").lower()
if ch == 'o':
    print("Outstanding")
elif ch=='a':
    print("Very Good")
elif ch=='b':
    print("Good")
elif ch=='c':
    print("Average")
elif ch =='f':
    print("fali")
else:
    print("invalide input")
```
