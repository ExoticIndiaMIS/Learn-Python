# Wap and modify the above program to calculate the parking charges. Read the hours and minutes when the vehicle enters the parking lot. When the vehical is leaving,enter its leaving time . Calculate the difference between the two timing to calculate the number of hours and minutes for which the vehical was parked Finally, calculate the charge based on following rules and then display the result on screen.


```python
print("Enter 'c' for Car, 'b' or 't' for Truck, and 'm' for Two-wheeler Vehicle.")
type = input("Enter the vehicle type: ")

e_hour = int(input("Enter the entry hour (0-23): "))
e_mint = int(input("Enter the entry minutes (0-59): "))
ex_hour = int(input("Enter the exit hour (0-23): "))
ex_mint = int(input("Enter the exit minutes (0-59): "))
entry_time = e_hour * 60 + e_mint
exit_time = ex_hour * 60 + ex_mint
if exit_time < entry_time:
    print("Error: Exit time cannot be before entry time.")
else:
    total_minutes = exit_time - entry_time
    l_hour = total_minutes // 60
    l_min = total_minutes % 60  
    price = 0
    if type == 't' or type == 'b':
        if l_hour <= 3:
            price = l_hour * 20
        else:
            price = 3 * 20 + (l_hour - 3) * 30  
    elif type == 'c':
        if l_hour <= 3:
            price = l_hour * 10
        else:
            price = 3 * 10 + (l_hour - 3) * 20  
    elif type == 'm':
        if l_hour <= 3:
            price = l_hour * 5
        else:
            price = 3 * 5 + (l_hour - 3) * 10  # Corrected formula
    else:
        print("Enter the correct type of vehicle")
        price = None
    if price is not None:
        print(f"For vehicle type '{type}', parking charge is ₹{price}")
```
