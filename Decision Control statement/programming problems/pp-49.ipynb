# A video library rents new videos for $75 a day, and old movies for $ 50 a day. Write a program to calculate the totyal charge for a customer's videos rental.The program should prompt the user for the number of each type of videos and output the total cost.

```python
# Prompt the user to enter the number of new and old videos rented
new_videos = int(input("Enter the number of new videos rented: "))
old_videos = int(input("Enter the number of old videos rented: "))

# Define rental prices
new_video_price = 75
old_video_price = 50

# Calculate total rental cost
total_cost = (new_videos * new_video_price) + (old_videos * old_video_price)

# Display the total cost
print(f"Total rental cost: ${total_cost}")
```
