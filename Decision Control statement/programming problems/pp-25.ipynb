# Wap to display the cos(x) and tan(x) value where range from 0 to 360 in steps of 15.

```python
import math
print("x (degrees)  cos(x)    tan(x)")
print("-------------------------------")
for x in range(0, 361, 15):  
    cos_x = round(math.cos(math.radians(x)), 4)  
    tan_x = round(math.tan(math.radians(x)), 4)
    print(f"{x:<12} {cos_x:<10} {tan_x}")
```
