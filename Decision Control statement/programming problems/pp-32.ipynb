# Wap to print a table of sine and cos funcrtion for the interval from 0-360 degrees in increments of 15.

```python
import math
print("Degree      sin(x)      cos(x)")
print("------------------------------")
for x in range (0,360,15):
    sin_x = round(math.sin(math.radians(x)), 4)  
    cos_x = round(math.cos(math.radians(x)), 4)
    print(f"{x:<12} {sin_x:<10} {cos_x}")
```
