# Wap to calculate electricity bill based on the following information.
# Consumption Unit    Rate of charge 
# 0-150               #$3 per unit
# 151-350             #$3.75 per unit
# 351-450             #$4 per unit
# 451-600             #$4.25 per unit
# above 600           #$5 per unit 

```python
unit = float(input("enter the consumption unit="))
rate = 0
if unit>=0 and unit <=150 :
    rate = unit*3
elif unit>=151 and unit<=350:
    rate = unit*3.75
elif unit>=351 and unit<=450:
    rate = unit*4
elif unit>=451 and unit <=600:
    rate = unit*4.25
else:
    rate = unit*5
print(f"electricity bill amonut for the {unit} unit is {rate}")
    
```
