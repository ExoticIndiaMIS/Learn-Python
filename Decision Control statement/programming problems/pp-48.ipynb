# Wap to read 5 digit number and then displat the number in the following formates for example the if the user entered 12345 the result should be .

```python
n = 5  # Number of rows

for i in range(n):  
    print(" " * i, end="")  # Print leading spaces
    for j in range(i+1 , n + 1):  # Print numbers from i+1 to n
        print(j, end="")
    print()  # Move to the next line
```
