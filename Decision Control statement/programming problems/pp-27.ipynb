# Wap to read an angle from the user and then display its quadrant.

```python
ang = int(input("enter the angle ="))
if ang>=0 and ang<=90:
    print(f"{ang} belong to first quadrant")
elif ang>=91 and ang<=180:
    print(f"{ang} belong to second quadrant")
elif ang>=181 and ang<=270:
    print(f"{ang} belong to third quadrant")
elif ang>=271 and ang <= 360:
    print(f"{ang} belong to fourth quadrant")
else:
    print(f"{ang} is not in the range of 0 to 360 ")
    
```
