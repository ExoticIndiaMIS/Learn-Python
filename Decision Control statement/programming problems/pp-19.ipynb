# Enter two integer as dividend and divisor. If the divisor is grater than zero then divide the dividend by the divisor. Assign their result to an integer variable rem and their quotient to floating point number quo.

```python
dividend = int(input("enter the value of dividend="))
divisor = int(input("enter the value of divisor="))
rem = 0
quo = 0
if divisor>0:
    rem = int(dividend%divisor)
    quo = float(dividend/divisor)
else:
    print("divisor should be grater than o")
print("Reminder=",rem)
print("Quotient=",quo)
    
```
