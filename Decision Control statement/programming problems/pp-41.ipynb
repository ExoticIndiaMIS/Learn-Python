# Wap to generate the following pattern.

```python
rows = 6  # Number of rows
cols = 6  # Number of columns

for i in range(rows):
    for j in range(cols):
        if (i == 0 and j == 0)or (i == rows - 1 and j == cols - 1):  
            print("$", end=" ")  # Print $ at top-left and bottom-right
        elif i == 0 or i == rows - 1:  
            print("*", end=" ")  # Print * at first and last row
        elif j == i :  
            print("$", end=" ")  # Print $ in diagonal
        elif j == 0 or j==cols-1:  
            print("*", end=" ")  # Print * at the first column
        else:
            print(" ", end=" ")  # Print spaces inside the pattern
    print()  # Move to the next line
```
