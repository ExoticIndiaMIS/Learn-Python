# Wap to read a floating point number and a integer . if the value of the floating point number is grater than 4.14, then add 10 to the integer.

```python
f = float(input("enter the floating number ="))
i = int(input("enter the integer number ="))
if f>4.14:
    i = i+10
    f = f
else : # it can be used without the else block .
    i = i
    f = f
print(f"integer value {i}")
print(f"float value {f}")
```
